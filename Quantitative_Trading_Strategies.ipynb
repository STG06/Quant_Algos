## Trend Following Techniques

## Moving Averages
Simple n-day moving average =  (1/N)∑Closex

Exponential moving average 
$$
x_t = \alpha \,\text{close}_t + (1 - \alpha)\, x_{t-1}
$$

where the smoothing parameter $\alpha$ (for an EMA with a given number of days) is:

$$
\alpha = \frac{2}{1 + \text{days}}
$$

 weighted moving average assigns larger weights to more recent prices.

For a 4-period WMA:

$$
\text{WMA}_t
= \frac{1}{4 + 3 + 2 + 1}
\left(
4 \cdot \text{close}_t
+ 3 \cdot \text{close}_{t-1}
+ 2 \cdot \text{close}_{t-2}
+ 1 \cdot \text{close}_{t-3}
\right)
$$

Moving Average Rules:
Variation 1:
1. Buy when close crosses above x day moving average
2. Sell when close crosses below x day moving average

Variation 2:
1. Buy when x day moving average crosses above y day moving average
2. Sell when x day moving average crosses below y day moving average

## Channel Breakouts
A running tally of the highest highs and lowest lows over a fixed interval of days. A 20-day channel involves calculating and plotting the highest and lowest close over the past 20 days. As new price highs and lows are made, the channel moves higher and lower, contracting and expanding with market volatility.

Channel Breakout Rules:
1. Buy if today's close is the highest close of the past x days
2. Sell if today's close is the lowest close of the past x days
---

## Momentum

$$
\text{Momentum}_t = \text{Value}_t - \text{Value}_{t-\text{days}}
$$

Momentum Rules:
1. Buy when today's close is greater than the close x days ago
2. Sell when today's close is less than the close x days ago

---

## Volatility Breakouts
Short-term trading systems that enter positions after large price moves, based on the idea that big moves tend to be followed by further movement in the same direction

The Strategy consists of 3 components:
1. Reference Value
2. Volatility Measure
3. Volatility Multiplier


Reference Value

The reference value defines the price level from which movement is measured.

Common choices:
- Previous day’s close
- Today’s opening price
- Short-term moving average of closes


## 2. Volatility Measures

Volatility is used to distinguish **significant moves** from random price changes.

Common volatility measures:

### (a) Standard Deviation of Returns
- Measures dispersion of returns
- Most statistically interpretable
- Common proxy for market volatility

### (b) Average True Range (ATR)
Accounts for price gaps and limit moves.

True Range (TR) is the maximum of:
- Today’s high − today’s low
- Today’s high − yesterday’s close
- Yesterday’s close − today’s low

ATR is the moving average of TR over time.

### (c) Standard Deviation of Prices
- Measures dispersion of price levels
- Less theoretically clean, but effective in practice

Despite differences, these measures tend to produce **similar volatility estimates**.


## 3. Volatility Multiplier

The volatility multiplier controls **signal sensitivity**:
- Larger multiplier → fewer trades, stronger signals
- Smaller multiplier → more trades, noisier signals


## Entry Rules

Let:
- `Ref` = reference value
- `Vol` = volatility measure
- `k` = volatility multiplier

Then:

**Upper trigger (buy):**
\[
\text{Ref} + k \cdot \text{Vol}
\]

**Lower trigger (sell):**
\[
\text{Ref} - k \cdot \text{Vol}
\]

- **Buy** when price closes above the upper trigger
- **Sell / short** when price closes below the lower trigger

## Price Oscillators

## Relative Strength Index
Market tops are completed when the indicator rises above 70, while bottoms are formed during periods when the indicator falls below 30. 

Today, most people prepare to buy when the market on dips below 30.

$$
\text{RSI} = 100 - \frac{100}{1 + \frac{U}{D}}
$$

where:
- $U$ = average **upward** price change over the look-back period  
- $D$ = average **downward** price change over the look-back period  

RSI ranges from **0 to 100**.

--- 

## Stochastics
The stochastic oscillator compares the current closing price to the recent
high–low range over a look-back period.


## Raw %K

$$
\text{Raw } \%K
= \frac{\text{Close}_t - \text{Lowest Low}}
{\text{Highest High} - \text{Lowest Low}}
$$

where:
- **Highest High** = highest price over the look-back period  
- **Lowest Low** = lowest price over the look-back period  

The raw %K ranges from **0 to 1** (often scaled to **0–100**).


## Fast %K and Fast %D

- **Fast %K** is obtained by smoothing Raw %K with a **3-day moving average**
- **Fast %D** is obtained by smoothing Fast %K with another **3-day moving average**


## Trading Interpretation

- **Long entry**:
  - Fast %K rises above **30**
  - Fast %K crosses **above** Fast %D

- **Short entry**:
  - Fast %K falls below **80**
  - Fast %K crosses **below** Fast %D

## Moving Average Convergence/Divergence
The **Moving Average Convergence/Divergence (MACD)** is a momentum oscillator
developed by **Gerald Appel (1980s)**. It measures momentum by taking the
difference between two exponential moving averages (EMAs).


### MACD Line

$$
\text{MACD}_t
= \text{EMA}_{12}(\text{Close}_t)
- \text{EMA}_{26}(\text{Close}_t)
$$

- The **12-day EMA** responds faster to price changes  
- The **26-day EMA** responds more slowly  

As prices rise, the fast EMA increases faster than the slow EMA, causing MACD to
increase. In declining markets, MACD decreases.


### Signal Line

$$
\text{Signal}_t = \text{EMA}_{9}(\text{MACD}_t)
$$

The signal line smooths the MACD and is used to generate trading signals.


## Trading Interpretation

MACD signals are generated from **crossovers** between the MACD line and the
signal line.

### Bearish Signal (Short Entry)
- MACD rallies and stalls
- MACD crosses **below** the signal line  
- Indicates **trend exhaustion**

### Bullish Signal (Long Entry)
- MACD declines and trades flat
- MACD crosses **above** the signal line  
- Indicates **oversold conditions**

## Evaluating Trading Strategy 


### Sharpe Ratio
The ratio is calculated using mean and standard deviation. The numerator is calculated by averagign returns over time using daily, weekly, or monthly returns. For each period's return, we subtract the return of a risk-free instrument such as short-term Treasury bills. The denominator is the standard deviation of returns.

Sharpe Ratio = Average return / Standard deviation of returns * Scaling Factor

Scaling factor is equal to the square root of time periods in a year: sqrt(252).

--- 

### K-ratio: Formula and Interpretation

The **K-ratio** measures the **stability of a strategy’s equity curve over time**.
Unlike the Sharpe ratio, it explicitly accounts for *when* profits occur.



### K-ratio Formula

$$
\text{K-ratio}
= \frac{b_1}{\sigma_{b_1} \cdot \text{Obs}}
$$



## Variable Definitions

### $b_1$ — Regression Slope (Reward)
- Slope of a linear regression fitted to the equity curve
- Measures **average growth per period**
- Larger $b_1$ ⇒ faster equity growth


### $\sigma_{b_1}$ — Standard Error of the Slope (Risk)
- Measures how **stable** the equity growth is
- Large values indicate:
  - Choppy performance
  - Gains concentrated in a few trades
- Small values indicate:
  - Smooth, consistent returns



### $\text{Obs}$ — Number of Observations
- Total number of time periods in the test
- Used to **normalize** the K-ratio so results are comparable across:
  - Different timeframes
  - Different data frequencies



## How to Interpret the K-ratio

The K-ratio increases when:
- Equity grows faster (**higher $b_1$**)
- Equity growth is smoother (**lower $\sigma_{b_1}$**)
- Results persist over time (**more observations**)


## What Is a “Good” K-ratio?

Rule of thumb:

| K-ratio | Interpretation |
|--------|----------------|
| < 0.00 | Losing or unstable |
| 0.00 – 0.30 | Weak / unreliable |
| 0.30 – 0.50 | Acceptable |
| **> 0.50** | **Strong strategy** |
| > 1.00 | Very stable, high quality |


## Popular Trading Strategies 

### Channel Breakout
Enters long when a market's close is the highest close of the past 40 days, and enters short when a market's close is the lowest close of the past 40 days. 

Longs are exited if today's close is the lowest close of the past 20 days. Shorts are exited if today's close is the highest close of the past 20 days. 

### Dual Moving Average Crossover
Use two moving averages for creating trading signals. A 10-day and 40-day simple moving average are calculated every day. Long entries are established if the 10-day average crosses above the 40-day average. Short entries are established if the 10-day evergae crosses below the 40-day average. Positions are exited on the entry signal in the opposite direction. 

### Momentum 
Two parameter: 20-day momentum and a longer 80-day momentum. 

20-day momentum: long enteries are established if today's close is less than the close 20 days ago

80-day momentum: long entries are established if today's close is greater than the close 80 days ago. Short entries are established if today's close is less than the close 80 days ago.

### Volatility Breakout
Enter long if today's price change is greater than twice the standard deviation of price changes over the past 100 days. Enter short if today's price change is less than negative two times the standard deviation of price changes over the past 100 days. 

### Relative Strength Index
Long entries are established if today's 14-day RSI falls below 35. Short entries are established if today's 14-day RSI rises above 65.

### Moving Average Convergence/Divergence
The MACD is constructed by calculating the difference between a 12-day and 26-day exponential moving average. The signal line is generated by calculating a nine-day exponential moving average of the MACD. Long entries are established when the MACD rises above the signal line. Short entries are established when the MACD falls below the signal line.

# Testing Pairs for Tradability (Cointegration + Mean Reversion)

## Big idea: cointegration vs. tradability
A candidate stock pair is **ideally tradable** if the two stocks are **cointegrated**. In practice, pairs are rarely *perfectly* cointegrated: residual factor exposure can cause **mean drift**, making the spread (signal) not perfectly stationary.  
However, if the **signal-to-noise ratio** is strong enough, we can treat the residual as “stationary enough” over the trading horizon.

So the real question becomes:

> **How do we decide a pair is tradable even if it deviates from ideal cointegration?**

---

## Why “high correlation / good regression fit” is not enough
Cointegrated systems share a **common trend** (Stock–Watson view: each series = trend + stationary component; cointegration implies the trends match).

It is tempting to say:
- “If regressing one price series on the other gives high $$R^2$$ and strong t-stats, they must be cointegrated.”

But this is **not a valid test**, because even **independent random walks** can produce high $$R^2$$ in regression (Granger–Newbold **spurious regression**).  
So **strong correlation / high $$R^2$$ is not exclusive to cointegration**.

---

## Engle–Granger cointegration test (2-step)
The defining property of cointegration is:

> There exists a **linear combination** of the two series that is **stationary** (mean-reverting).

Engle–Granger procedure:

1. **Estimate the linear relationship** (equilibrium relationship) via regression.
2. **Test the residual** (spread) for **stationarity**.

If the residual is stationary → the pair is cointegrated.

---

## Tradability testing (practical adaptation)
For **tradability**, we still do **two steps**, but step 2 changes:

1. **Estimate the linear relationship** (same as cointegration testing).
2. Instead of strict stationarity, require the residual to show a **sufficient degree of mean reversion**.

So tradability = “spread mean-reverts enough to trade,” even if not perfectly stationary.

---

## The linear (equilibrium) relationship

The equilibrium relationship is modeled as:

$$
\log p^A_t - \gamma \log p^B_t = \mu + \varepsilon_t
$$

- $\gamma$ = cointegration coefficient (hedge ratio / common-factor beta)  
- $\mu$ = equilibrium value / premium  
- $\varepsilon_t$ = disturbance term (zero-mean deviations around equilibrium)

**Economic interpretation:**
- Consider a portfolio: **long 1 share of A**, **short $$\gamma$$ shares of B**.  
- The equilibrium relationship implies the portfolio has an average “cashflow” / premium of $$\mu$$.
- Premiums can exist due to liquidity premium, takeover premium, etc., so estimating $$\mu$$ matters.

---

## Estimating $$\gamma$$ and $$\mu$$

### 1) Multifactor approach
Using multifactor constructs, $$\gamma$$ can be written via covariance/variance and then expressed in factor exposures and factor covariance matrix.

If B is independent variable:

$$
\gamma = \beta_{AB} = \frac{\operatorname{cov}(r_A, r_B)}{\operatorname{var}(r_B)}
$$

If A is independent variable:

$$
\gamma' = \beta_{BA} = \frac{\operatorname{cov}(r_A, r_B)}{\operatorname{var}(r_A)}
$$

**Which one to use?**  
Use the **larger** of $$\gamma$$ and $$\gamma'$$ (numerically more stable, and it effectively makes the **lower-volatility** stock the independent variable).

Then:
- Build residual series from the chosen linear combo
- Estimate $$\mu$$ as the **mean** of the residual; if not significantly different from 0, treat $$\mu \approx 0$$.

---

### 2) Regression approach (OLS) + why it’s tricky
In standard regression (physical “stimulus → response”), we often assume:
- negligible error in the independent variable
- constant error variance in the dependent variable

Stock prices violate this:
- both price series have measurement uncertainty (e.g., within-day range)
- error variance can change over time (heteroskedasticity)
- cause/effect between two stocks isn’t clearly directional

A more “correct” approach is **weighted least squares** / chi-squared minimization:

$$
\chi^2(\gamma, \mu) = \sum_t \frac{\left(\log p^A_t - \gamma \log p^B_t - \mu\right)^2}{\operatorname{var}_t(\log p^A_t) + \gamma^2 \operatorname{var}_t(\log p^B_t)}
$$

But the $$\gamma^2$$ term in the denominator makes optimization nonlinear and potentially requires numerical methods.

**Practical simplification:** use **VWAP** prices for regression:
- VWAP is a proxy for “maximum-liquidity” consensus price
- makes error distributions more uniform across time
- reduces sensitivity to outliers
- allows using simpler OLS in practice (t-stats less emphasized in this application)

**Note:** even if regression uses VWAP, residual series can still be constructed using close-close data to visualize oscillations around equilibrium.

---

## Testing residuals for tradability via mean reversion
Pairs trading bets that the residual (spread) **reverts to its mean** $$\mu$$.

A useful, model-free proxy for mean reversion:
- **Zero-crossing frequency** (how often residual crosses its long-run mean)

High zero-crossing rate ⇒
- fast mean reversion
- shorter holding time
- better signal-to-noise
- stronger case for tradability

Contrast: **Brownian motion** (nonstationary) has **infrequent** zero crossings; related intuition comes from Lévy’s **arcsine law** (last time at zero tends to occur near the endpoints of the interval).

---

## Estimating zero-crossing behavior (bootstrap, model-free)
Directly counting crossings on a single residual sample is biased (we only have one realization).  
Instead, estimate the distribution of **time between crossings** (inverse of crossing rate) using **bootstrap**:

1. Compute sample times between successive zero crossings from the residual series.
2. Resample (bootstrap) these intervals many times to build an empirical distribution.
3. Use percentiles (around the median) to check if holding periods meet trading requirements.  
   If yes → declare the pair tradable.

Bonus: the time-between-crossings distribution provides a **holding period benchmark** and can inform **time-based stops**.

---

## Example outcome (illustrative)
After estimating the equilibrium relationship:

- $$\mu = -0.6971$$
- $$\gamma = 1.0617$$
- Regression $$R^2 = 0.7965$$

Residual plot looks tradable; bootstrap of time-between-crossings suggests:
- extremes around ~14–26 days
- median around ~5 days

Interpretation: tradable pair; may be preferable to exit on reversion to zero rather than waiting for a large excursion in the opposite direction.


# Trading Signal / Band Design for Pairs Trading (Summary)

## Core setup
- Pairs trading assumes an **equilibrium relationship** between two stocks’ **log prices**, defined by:
  - **cointegration coefficient** (hedge ratio)
  - **equilibrium value** (mean level)
- Using these, we form the **spread** = a linear combination of the two log-price series.
- The trade is a bet on **mean reversion** of the spread:
  - **enter** when the spread deviates “enough” from equilibrium
  - **exit** when it reverts back
- Therefore, we must explicitly define a divergence threshold **$\Delta$** → this becomes a clear set of **trading signals** (rules).

---

## Spread dynamics motivation
- Spreads are chosen to be mean-reverting, but their dynamics can vary widely.
- A convenient assumption is that spreads are **stationary ARMA processes**, meaning many different (but mean-reverting) behaviors are possible.
- Goal: design trading rules that **maximize profits**.
- Instead of building different rules for each spread type, the chapter argues for a **“one-size-fits-all”** methodology.

---

## Band design for Gaussian white-noise spreads
### Trading rule example
- **Buy** 1 unit of spread when spread $\le -\Delta$
- **Sell** 1 unit of spread when spread $\ge \Delta$

For a Gaussian white-noise spread, the tail probability is:
- $P(\text{spread} \ge \Delta) = 1 - N(\Delta)$
- By symmetry, $P(\text{spread} \le -\Delta) = 1 - N(\Delta)$

Over $T$ time steps:
- Expected # of “sell” triggers $\approx T(1 - N(\Delta))$
- Expected # of “buy” triggers $\approx T(1 - N(\Delta))$
- Expected total trades $\approx 2T(1 - N(\Delta))$

Profit per completed buy+sell cycle is about:
- $$2\Delta$$

So a simple profitability proxy is proportional to:
$$
\Delta \, (1 - N(\Delta))
$$

**Optimal threshold:** the profit proxy peaks at about:
$$
\Delta \approx 0.75\sigma
$$

### Inventory constraint insight
Even if you restrict inventory to **one spread unit** (flip position instead of adding), simulations show the **same optimal $\Delta$**.  
So **inventory constraints do not change the optimal band level** in this white-noise case.

---

## Limitations of the simple white-noise band design
- Assumes **uniform liquidity** at all divergence levels (not realistic).
- Says **nothing about position sizing** (in reality, market impact/slippage limits how large you can trade).
- Best viewed as a **guideline** when you lack better information.

---

## Why parametric modeling gets complicated
As you try to model spreads more realistically, models quickly escalate in complexity:

1. **Mixture Gaussian / time-varying volatility**
   - volatility higher near open/close; possibly GARCH-like
   - may require **multiple thresholds** or **time-varying bands** (e.g., dynamic volatility estimates)

2. **ARMA spreads**
   - spreads can show autocorrelation/momentum (e.g., after news events)
   - if you can compute level-crossing rates (e.g., via Rice’s formula), you can optimize $\Delta$

3. **Hidden Markov / regime-switching ARMA**
   - volatility regimes chosen by a Markov process; parameters estimated via algorithms like Baum–Welch
   - level-crossing rates often require **simulation** to optimize thresholds

**Takeaway:** fully parametric approaches can become computationally heavy.

---

## Nonparametric approach (preferred)
Goal: choose $\Delta$ **without** fitting a full parametric model of spread dynamics.

Idea: estimate the **profit profile** directly from recent spread data.

Key practical points:
- Use **discrete candidate thresholds** (levels closer than slippage are indistinguishable).
- Estimate how often the spread crosses each threshold (counts).
- Multiply crossing frequency by profit-per-trade → **raw profit curve** (often noisy with small samples).
- Fix issues in two steps:
  1. Enforce **monotonicity** of the crossing-rate curve (correct discretization step effects)
  2. Apply **Tikhonov–Miller regularization** (smooth the function)

Regularization trades off fit vs smoothness by minimizing:
$$
\text{cost}=\sum_{i=1}^{n}(y_i-z_i)^2+\lambda\sum_{i=1}^{n-1}(z_i-z_{i+1})^2
$$

Choose $\lambda$ at the **“heel”** of the fit-error vs $\lambda$ curve (balance fit and smoothness).  
Then compute the smoothed profit profile and pick the $\Delta$ at its maximum.

Validation: applying this method to white noise recovers the theoretical optimum:
$$
\Delta \approx 0.75\sigma
$$

---

## Practical notes (“loose ends”)
- **Multiple thresholds** may be needed for mixture distributions (segment by time-of-day).
- Alternatively, estimate thresholds **dynamically** using a rolling window.
- **Time-based stops** matter because mean drift can degrade SNR over time; sometimes exit at mean/zero instead of waiting for an opposite-side excursion.

---

## Summary bullets
- Choose entry/exit bands $\Delta$ to **maximize profits** (tradeoff: frequency vs profit per trade).
- The optimal $\Delta$ lies between “too small” and “too large.”
- Nonparametric estimation of the profit profile is often simpler/more robust than heavy parametric modeling.
- A practical recipe: **monotonic adjustment → regularization → maximize profit profile** to choose $\Delta$.
