## Trend Following Techniques

## Moving Averages
Simple n-day moving average =  (1/N)∑Closex

Exponential moving average 
$$
x_t = \alpha \,\text{close}_t + (1 - \alpha)\, x_{t-1}
$$

where the smoothing parameter $\alpha$ (for an EMA with a given number of days) is:

$$
\alpha = \frac{2}{1 + \text{days}}
$$

 weighted moving average assigns larger weights to more recent prices.

For a 4-period WMA:

$$
\text{WMA}_t
= \frac{1}{4 + 3 + 2 + 1}
\left(
4 \cdot \text{close}_t
+ 3 \cdot \text{close}_{t-1}
+ 2 \cdot \text{close}_{t-2}
+ 1 \cdot \text{close}_{t-3}
\right)
$$

Moving Average Rules:
Variation 1:
1. Buy when close crosses above x day moving average
2. Sell when close crosses below x day moving average

Variation 2:
1. Buy when x day moving average crosses above y day moving average
2. Sell when x day moving average crosses below y day moving average

## Channel Breakouts
A running tally of the highest highs and lowest lows over a fixed interval of days. A 20-day channel involves calculating and plotting the highest and lowest close over the past 20 days. As new price highs and lows are made, the channel moves higher and lower, contracting and expanding with market volatility.

Channel Breakout Rules:
1. Buy if today's close is the highest close of the past x days
2. Sell if today's close is the lowest close of the past x days
---

## Momentum

$$
\text{Momentum}_t = \text{Value}_t - \text{Value}_{t-\text{days}}
$$

Momentum Rules:
1. Buy when today's close is greater than the close x days ago
2. Sell when today's close is less than the close x days ago

## Volatility Breakouts
Short-term trading systems that enter positions after large price moves, based on the idea that big moves tend to be followed by further movement in the same direction

The Strategy consists of 3 components:
1. Reference Value
2. Volatility Measure
3. Volatility Multiplier


Reference Value

The reference value defines the price level from which movement is measured.

Common choices:
- Previous day’s close
- Today’s opening price
- Short-term moving average of closes


## 2. Volatility Measures

Volatility is used to distinguish **significant moves** from random price changes.

Common volatility measures:

### (a) Standard Deviation of Returns
- Measures dispersion of returns
- Most statistically interpretable
- Common proxy for market volatility

### (b) Average True Range (ATR)
Accounts for price gaps and limit moves.

True Range (TR) is the maximum of:
- Today’s high − today’s low
- Today’s high − yesterday’s close
- Yesterday’s close − today’s low

ATR is the moving average of TR over time.

### (c) Standard Deviation of Prices
- Measures dispersion of price levels
- Less theoretically clean, but effective in practice

Despite differences, these measures tend to produce **similar volatility estimates**.


## 3. Volatility Multiplier

The volatility multiplier controls **signal sensitivity**:
- Larger multiplier → fewer trades, stronger signals
- Smaller multiplier → more trades, noisier signals


## Entry Rules

Let:
- `Ref` = reference value
- `Vol` = volatility measure
- `k` = volatility multiplier

Then:

**Upper trigger (buy):**
\[
\text{Ref} + k \cdot \text{Vol}
\]

**Lower trigger (sell):**
\[
\text{Ref} - k \cdot \text{Vol}
\]

- **Buy** when price closes above the upper trigger
- **Sell / short** when price closes below the lower trigger

## Price Oscillators

## Relative Strength Index
Market tops are completed when the indicator rises above 70, while bottoms are formed during periods when the indicator falls below 30. 

Today, most people prepare to buy when the market on dips below 30.

$$
\text{RSI} = 100 - \frac{100}{1 + \frac{U}{D}}
$$

where:
- $U$ = average **upward** price change over the look-back period  
- $D$ = average **downward** price change over the look-back period  

RSI ranges from **0 to 100**.

--- 

## Stochastics
The stochastic oscillator compares the current closing price to the recent
high–low range over a look-back period.


## Raw %K

$$
\text{Raw } \%K
= \frac{\text{Close}_t - \text{Lowest Low}}
{\text{Highest High} - \text{Lowest Low}}
$$

where:
- **Highest High** = highest price over the look-back period  
- **Lowest Low** = lowest price over the look-back period  

The raw %K ranges from **0 to 1** (often scaled to **0–100**).


## Fast %K and Fast %D

- **Fast %K** is obtained by smoothing Raw %K with a **3-day moving average**
- **Fast %D** is obtained by smoothing Fast %K with another **3-day moving average**


## Trading Interpretation

- **Long entry**:
  - Fast %K rises above **30**
  - Fast %K crosses **above** Fast %D

- **Short entry**:
  - Fast %K falls below **80**
  - Fast %K crosses **below** Fast %D

## Moving Average Convergence/Divergence
The **Moving Average Convergence/Divergence (MACD)** is a momentum oscillator
developed by **Gerald Appel (1980s)**. It measures momentum by taking the
difference between two exponential moving averages (EMAs).


### MACD Line

$$
\text{MACD}_t
= \text{EMA}_{12}(\text{Close}_t)
- \text{EMA}_{26}(\text{Close}_t)
$$

- The **12-day EMA** responds faster to price changes  
- The **26-day EMA** responds more slowly  

As prices rise, the fast EMA increases faster than the slow EMA, causing MACD to
increase. In declining markets, MACD decreases.


### Signal Line

$$
\text{Signal}_t = \text{EMA}_{9}(\text{MACD}_t)
$$

The signal line smooths the MACD and is used to generate trading signals.


## Trading Interpretation

MACD signals are generated from **crossovers** between the MACD line and the
signal line.

### Bearish Signal (Short Entry)
- MACD rallies and stalls
- MACD crosses **below** the signal line  
- Indicates **trend exhaustion**

### Bullish Signal (Long Entry)
- MACD declines and trades flat
- MACD crosses **above** the signal line  
- Indicates **oversold conditions**
