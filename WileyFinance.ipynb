## Time Series

### Autocorrelation
*Autocorrelation** measures the linear relationship between a time series and a
lagged version of itself.

It answers questions such as:
- How related is the current value to the value 1 time step ago?
- 2 time steps ago? 3 time steps ago?

---
Definition:

For a time series \(\{y_t\}_{t=1}^T\), the **autocorrelation at lag \(\tau\)** is the
correlation between \(y_t\) and \(y_{t-\tau}\).

---
 Sample Autocorrelation Formula
$$
\hat{\rho}(\tau)
=
\frac{
\sum_{t=\tau+1}^{T} (y_t - \bar{y})(y_{t-\tau} - \bar{y})
}{
\sum_{t=1}^{T} (y_t - \bar{y})^2
}
$$

where:
- $T$ = number of observations  
- $\tau$ = time lag  
- $y_t$ = value at time \(t\)  
- $\bar{y}$ = sample mean of the series


---

Interpretation

- $\hat{\rho}(\tau) \approx 0$: no linear dependence at lag \(\tau\)
- $\hat{\rho}(\tau) > 0$: persistence / trending behavior
- $\hat{\rho}(\tau) < 0$: mean reversion or oscillation

---
### White Noise
The variance of the value at each point in the series is the variance of the normal distribution used for drawing white noise

---

### Moving Average Process

A **moving average (MA) process** constructs a time series as a linear combination
of current and past white noise realizations.

It answers questions such as:
- Does past random noise affect the current value?
- For how many time steps does this dependence last?

---

Definition:

A **moving average process of order 1**, denoted **MA(1)**, is defined as:

$$
y_t = e_t + \beta e_{t-1}
$$

where $e_t$ is white noise and $\beta$ is a constant coefficient.

---

General MA(q) Formula

A moving average process of order $q$, denoted **MA(q)**, is defined as:

$$
y_t = e_t + b_1 e_{t-1} + b_2 e_{t-2} + \cdots + b_q e_{t-q}
$$

where:
- $e_t$ = white noise  
- $b_1, \dots, b_q$ = constants  
- $q$ = order of the MA process  

---

Interpretation

- MA(1) shows autocorrelation at **lag 1 only**
- MA(q) shows autocorrelation up to **lag $q$**
- Autocorrelation is **zero beyond lag $q$**
- Past noise improves short-term predictability
---

### Autoregressive (AR) Process

An **autoregressive (AR) process** constructs a time series using a linear
combination of **past values of the series itself** plus a white noise term.

It answers questions such as:
- How strongly does the past value influence the current value?
- How long does this dependence persist?

---

Definition:

A **first-order autoregressive process**, denoted **AR(1)**, is defined as:

$$
y_t = \alpha y_{t-1} + e_t
$$

where $e_t$ is white noise and $\alpha$ is a constant coefficient.

---

Autocorrelation Behavior

- Correlation exists at **all lags**
- Autocorrelation **decays gradually** as lag increases
- No sharp cutoff in the correlogram

---

Interpretation

- Past values directly influence future values
- Larger $|\alpha|$ implies stronger persistence
- Knowledge of past values improves predictability

### ARMA and Random Walk Processes

The **ARMA process** combines autoregressive (AR) and moving average (MA)
models to capture dependence on both past values and past noise.

It answers questions such as:
- How do past values and past shocks jointly affect the present?
- How flexible is the time series structure?

---

Definition:

An **ARMA(p, q)** process is defined as:

$$
y_t = \alpha_1 y_{t-1} + \cdots + \alpha_p y_{t-p}
      + e_t + \beta_1 e_{t-1} + \cdots + \beta_q e_{t-q}
$$

where:
- $y_t$ = time series value  
- $e_t$ = white noise  
- $p$ = AR order  
- $q$ = MA order  

---

Interpretation

- Combines persistence (AR) and shock effects (MA)
- Sum of independent ARMA processes is also ARMA
- Commonly used in practical time series modeling

---

### Random Walk Process

A **random walk** is a special ARMA process and is defined as an **AR(1)** model
with $\alpha = 1$.

---

Definition:

$$
y_t = y_{t-1} + e_t
$$

---

Interpretation

- Variance increases **linearly with time**
- No mean reversion → strong trending behavior
- Nonstationary time series
- Best prediction of next value is the **current value**
- Random walk is a **martingale**

## Akaike Information Criterion (AIC)

The **Akaike Information Criterion (AIC)** balances **goodness of fit** and
**model complexity** to prevent overfitting.

---

### AIC Formula

$$
\text{AIC}
=
n \log\!\left(\frac{1}{n}\sum_{i=1}^{n} e_i^2\right)
+ 2k
$$

**where:**
- \( n \) is the number of observations  
- \( e_i \) is the forecast (residual) error at time \( i \)  
- \( k \) is the number of model parameters  

---

### Interpretation

- The **first term** measures goodness of fit:
  
  $$
  n \log\!\left(\frac{1}{n}\sum_{i=1}^{n} e_i^2\right)
  $$
  
  Smaller errors imply a better fit.

- The **second term** penalizes model complexity:
  
  $$
  2k
  $$
  
  Each additional parameter increases AIC by 2.

---

### Model Selection Rule

- **Lower AIC is better**
- Only **relative AIC values** matter
- Adding a parameter is worthwhile **only if it reduces the first term enough to offset the +2 penalty**

---

### Key Takeaways

- AIC formalizes the **bias–variance tradeoff**
- More parameters do not necessarily improve a model
- Widely used for **forecasting and time series models** (AR, ARMA, ARIMA)

---

### Related Criteria

- **BIC / SIC**: impose a stronger penalty on complexity
- AIC emphasizes **predictive accuracy**


## Arbitrage Pricing Theory (APT)

**Arbitrage Pricing Theory (APT)** was introduced by **Stephen A. Ross (1976)**.
It models asset returns using multiple risk factors and is grounded in
**no-arbitrage arguments**, rather than market equilibrium assumptions.

---

## Core Idea

> **An asset’s risk and return are fully determined by its exposure to a set of common risk factors.**

Each factor contributes to the asset’s return in proportion to its sensitivity
(exposure) to that factor.

---

## Return Model

Let:
- \( \beta_1, \beta_2, \ldots, \beta_k \) be factor exposures  
- \( r_1, r_2, \ldots, r_k \) be factor returns  
- \( r_e \) be the idiosyncratic (specific) return  

The asset return is:

$$
r = \beta_1 r_1 + \beta_2 r_2 + \cdots + \beta_k r_k + r_e
$$

**Key assumption:**  
The specific return \( r_e \) is uncorrelated with:
- all factor returns
- the specific returns of other assets

---

## Measuring Risk (Variance)

Risk is measured as the **variance of returns**, following the
**Markowitz mean–variance framework**.

For a two-factor model:

$$
r = \beta_1 r_1 + \beta_2 r_2 + r_e
$$

The variance of return is:

$$
\mathrm{var}(r)
=
\beta^\top V \beta
+
\mathrm{var}(r_e)
$$

where:
- \( \beta \) is the vector of factor exposures  
- \( V \) is the **factor covariance matrix**  

---

## Variance Decomposition

Total risk splits into **common factor risk** and **specific risk**:

$$
\sigma_{\text{total}}^2
=
\sigma_{\text{common}}^2
+
\sigma_{\text{specific}}^2
$$

Equivalently (Pythagorean analogy):

$$
\sigma_{\text{total}}
=
\sqrt{
\sigma_{\text{common}}^2
+
\sigma_{\text{specific}}^2
}
$$

In practice, **most risk comes from common factors**.

---

## Factor Model Components

An APT model is fully specified by:

1. **Factor Exposure Matrix** \( X \)  
   - Dimensions: \( N \times k \)  
   - Rows = stocks, columns = factors  

2. **Factor Covariance Matrix** \( V \)  
   - Dimensions: \( k \times k \)  
   - Diagonal: factor variances  
   - Off-diagonal: factor covariances  

3. **Specific Variance Matrix** \( \Delta \)  
   - Dimensions: \( N \times N \)  
   - Diagonal only (assumed uncorrelated)

---

## Covariance Matrix Properties

The factor covariance matrix:
- is **symmetric**
- is **positive definite**
- captures correlations between factor returns

Covariance between two assets \( A \) and \( B \):

$$
\mathrm{cov}(r_A, r_B)
=
\beta_A^\top V \beta_B
$$

Full asset covariance matrix:

$$
\mathrm{CovMatrix}
=
X V X^\top
$$

This dramatically reduces computational complexity for large universes.

---

## Practical Considerations

- Covariance matrices are **estimated from historical data**
- Subject to **estimation error** and **data-mining bias**
- Assume **Gaussian returns** (fat tails not captured)
- Despite limitations, they remain central to risk modeling

---

## Example

Factor exposure for stock A (two-factor model):

$$
\beta_A = (0.5,\, 0.75)
$$

Specific variance:

$$
\sigma_{\text{specific}}^2 = 0.0123
$$

Common factor variance:

$$
\sigma_{\text{common}}^2 = 0.0901
$$

Total variance:

$$
\sigma_{\text{total}}^2 = 0.0901 + 0.0123 = 0.1024
$$

Standard deviation (volatility):

$$
\sigma = \sqrt{0.1024} = 0.32
$$

**Stock A volatility = 32%**

---

## Key Takeaways

- APT explains returns via **multiple systematic factors**
- Risk decomposes into **factor risk + specific risk**
- Factor covariance matrix is the core risk input
- Enables scalable risk modeling for large asset universes

## Limitations of APT / Factor Risk Models

Factor models rely on **stable inputs**. Risk estimates can fail when these assumptions break.

---

### 1. Factor Exposure Instability (β Breakdown)

- Company-specific news (earnings, M&A, lawsuits) can instantly change a stock’s factor exposures
- Historical betas become invalid
- Risk estimates fail **for that stock**

---

### 2. Covariance Matrix Instability (V Breakdown)

- Macro or liquidity shocks can alter factor correlations
- During crises, correlations often spike toward 1
- The factor covariance matrix becomes unreliable
- Risk estimates fail **system-wide**

---

### 3. Non-Gaussian Returns (Fat Tails)

- Factor models assume approximately normal returns
- Real financial returns exhibit fat tails
- Large price moves occur more often than \( \pm 2\sigma \)
- Variance-based risk underestimates extreme events

---

### Key Takeaway

> APT is powerful but only as reliable as its factor exposures, covariance matrix,
and distributional assumptions—each of which can fail during market stress.

## Beta as a Hedge Ratio (CAPM within APT)

Although APT is more general than CAPM, **beta remains useful** when the goal
is to **hedge a portfolio with the market portfolio**.

In this context, **beta is the optimal hedge ratio**.

---

### Hedge Ratio Interpretation

Consider a combined position:
- Long the portfolio
- Short \( \lambda \) units of the market

Return of the combined position:

$$
r_p - \lambda r_m
$$

The **optimal hedge ratio** \( \lambda \) minimizes the variance of this return.

---

### Optimal Hedge Ratio

Minimizing variance yields:

$$
\lambda
=
\frac{\mathrm{cov}(r_p, r_m)}{\mathrm{var}(r_m)}
$$

This is **exactly the definition of beta**.

\[
\Rightarrow \quad \text{Beta } = \text{ best hedge ratio}
\]

Using APT, the covariance and variance terms are computed from:
- factor exposures
- factor covariance matrix
- specific variances

---

## Application: Tracking Basket Design

A **tracking basket** is a portfolio designed to replicate an index using
fewer securities.

Because the basket differs from the index, it exhibits **tracking error**.

---

### Tracking Error Definition

Tracking error is the **standard deviation of the return difference**
between the basket and the index:

$$
\text{Tracking Error}
=
\sqrt{\mathrm{var}(r_p - r_m)}
$$

Assuming zero mean difference, this measures **how tightly the basket tracks
the index**.

---

### Tracking Error Minimization

Tracking basket design solves:

$$
\min \; \mathrm{var}(r_p - r_m)
$$

Using APT, this variance decomposes into:
- **common factor variance**
- **specific (idiosyncratic) variance**

---

### Key Insight

If the tracking basket is constructed so that:

$$
\beta_p = \beta_m
$$

(i.e., factor exposures match the index)

then:
- **common factor risk is eliminated**
- tracking error comes only from specific risk

Since specific risk is smaller and diversifies away, **matching factor
exposures is the primary design goal**.

---

## Key Takeaways

- Beta is the **optimal hedge ratio** against the market
- Tracking error measures deviation from index returns
- APT enables efficient tracking basket design
- Matching **factor exposures** minimizes tracking error


## Kalman Filtering — Core Concepts

The **Kalman filter** is a recursive estimation algorithm that combines
**model predictions** and **noisy observations** to estimate the true
state of a system with **minimum error variance**.

It operates in a repeating **three-step cycle**:
1. **Prediction**
2. **Observation**
3. **Correction (Reconciliation)**

---

## 1. Prediction Step

Predict the next state using a linear state model:

$$
X_t = A X_{t-1} + u_t
$$

where:
- \( X_t \) is the system state
- \( A \) is the state transition matrix
- \( u_t \) is model (process) noise

Also estimate the **prediction error variance**.

---

## 2. Observation Step

Observe the system through a linear measurement model:

$$
Y_t = H X_t + v_t
$$

where:
- \( Y_t \) is the observation
- \( H \) maps the state to observations
- \( v_t \) is observation noise

Estimate the **observation error variance** \( R \).

---

## 3. Correction Step

Combine prediction and observation:

$$
\hat{X}_t
=
\hat{X}_{t|t-1}
+
K_t \bigl( Y_t - \hat{Y}_t \bigr)
$$

where:
- \( Y_t - \hat{Y}_t \) is the **innovation**
- \( K_t \) is the **Kalman gain**

The Kalman gain determines how much weight is given to the observation
versus the prediction.

---

## Kalman Gain (Scalar Case)

For a constant state with:
- prediction variance \( \sigma_\varepsilon^2 \)
- observation variance \( \sigma_\eta^2 \)

the optimal Kalman gain is:

$$
K
=
\frac{\sigma_\varepsilon^2}
{\sigma_\varepsilon^2 + \sigma_\eta^2}
$$

This makes the corrected estimate a **weighted average**:

$$
\hat{x}
=
(1-K)\,\hat{x}_{\text{pred}} + K\,y
$$

Lower observation noise ⇒ larger \( K \).

---

## Interpretation

- Kalman filtering generalizes **weighted averaging**
- More reliable measurements receive higher weight
- The corrected estimate has **minimum variance**

Kalman proved optimality when:
- models are linear
- errors are independent and Gaussian

---

## Kalman Filter vs Least Squares

- Solving all observations at once ⇒ **least squares**
- Kalman filter ⇒ **recursive least squares**
- Same solution, but **constant computational cost per step**

---

## Filtering a Random Walk

Random walk model:

$$
x_t = x_{t-1} + \varepsilon_t
$$

Observation:

$$
y_t = x_t + \eta_t
$$

This is relevant for finance because:
- log stock prices are often modeled as random walks

Kalman filtering estimates the **latent equilibrium price**, not just
the last observed price.

---

## Fibonacci Weights (Special Case)

When:
- state variance = observation variance
- errors are i.i.d.

The Kalman estimate becomes a **weighted sum of past observations**:

$$
x_T = w_0 y_T + w_1 y_{T-1} + w_2 y_{T-2} + \cdots
$$

Weights are ratios of **Fibonacci numbers**:

$$
w_0 \approx \frac{1}{\varphi} \approx 0.618
$$

where:

$$
\varphi = \frac{1+\sqrt{5}}{2}
$$

Older observations decay rapidly in importance.

---

## Practical Insight

- Kalman filtering is an **adaptive moving average**
- Observation and state variances control smoothness
- Sampling a random walk at lower frequency increases smoothing
- Widely used for **price smoothing, signal extraction, and state estimation**

---

## Key Takeaway

> The Kalman filter is a recursive, optimal way to combine model predictions
and noisy observations, generalizing moving averages and least squares into
a single dynamic framework.
