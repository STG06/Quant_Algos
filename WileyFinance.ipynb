## Time Series

### Autocorrelation
*Autocorrelation** measures the linear relationship between a time series and a
lagged version of itself.

It answers questions such as:
- How related is the current value to the value 1 time step ago?
- 2 time steps ago? 3 time steps ago?

---
Definition:

For a time series \(\{y_t\}_{t=1}^T\), the **autocorrelation at lag \(\tau\)** is the
correlation between \(y_t\) and \(y_{t-\tau}\).

---
 Sample Autocorrelation Formula
$$
\hat{\rho}(\tau)
=
\frac{
\sum_{t=\tau+1}^{T} (y_t - \bar{y})(y_{t-\tau} - \bar{y})
}{
\sum_{t=1}^{T} (y_t - \bar{y})^2
}
$$

where:
- $T$ = number of observations  
- $\tau$ = time lag  
- $y_t$ = value at time \(t\)  
- $\bar{y}$ = sample mean of the series


---

Interpretation

- $\hat{\rho}(\tau) \approx 0$: no linear dependence at lag \(\tau\)
- $\hat{\rho}(\tau) > 0$: persistence / trending behavior
- $\hat{\rho}(\tau) < 0$: mean reversion or oscillation

---
### White Noise
The variance of the value at each point in the series is the variance of the normal distribution used for drawing white noise

---

### Moving Average Process

A **moving average (MA) process** constructs a time series as a linear combination
of current and past white noise realizations.

It answers questions such as:
- Does past random noise affect the current value?
- For how many time steps does this dependence last?

---

Definition:

A **moving average process of order 1**, denoted **MA(1)**, is defined as:

$$
y_t = e_t + \beta e_{t-1}
$$

where $e_t$ is white noise and $\beta$ is a constant coefficient.

---

General MA(q) Formula

A moving average process of order $q$, denoted **MA(q)**, is defined as:

$$
y_t = e_t + b_1 e_{t-1} + b_2 e_{t-2} + \cdots + b_q e_{t-q}
$$

where:
- $e_t$ = white noise  
- $b_1, \dots, b_q$ = constants  
- $q$ = order of the MA process  

---

Interpretation

- MA(1) shows autocorrelation at **lag 1 only**
- MA(q) shows autocorrelation up to **lag $q$**
- Autocorrelation is **zero beyond lag $q$**
- Past noise improves short-term predictability
---

### Autoregressive (AR) Process

An **autoregressive (AR) process** constructs a time series using a linear
combination of **past values of the series itself** plus a white noise term.

It answers questions such as:
- How strongly does the past value influence the current value?
- How long does this dependence persist?

---

Definition:

A **first-order autoregressive process**, denoted **AR(1)**, is defined as:

$$
y_t = \alpha y_{t-1} + e_t
$$

where $e_t$ is white noise and $\alpha$ is a constant coefficient.

---

Autocorrelation Behavior

- Correlation exists at **all lags**
- Autocorrelation **decays gradually** as lag increases
- No sharp cutoff in the correlogram

---

Interpretation

- Past values directly influence future values
- Larger $|\alpha|$ implies stronger persistence
- Knowledge of past values improves predictability

### ARMA and Random Walk Processes

The **ARMA process** combines autoregressive (AR) and moving average (MA)
models to capture dependence on both past values and past noise.

It answers questions such as:
- How do past values and past shocks jointly affect the present?
- How flexible is the time series structure?

---

Definition:

An **ARMA(p, q)** process is defined as:

$$
y_t = \alpha_1 y_{t-1} + \cdots + \alpha_p y_{t-p}
      + e_t + \beta_1 e_{t-1} + \cdots + \beta_q e_{t-q}
$$

where:
- $y_t$ = time series value  
- $e_t$ = white noise  
- $p$ = AR order  
- $q$ = MA order  

---

Interpretation

- Combines persistence (AR) and shock effects (MA)
- Sum of independent ARMA processes is also ARMA
- Commonly used in practical time series modeling

---

### Random Walk Process

A **random walk** is a special ARMA process and is defined as an **AR(1)** model
with $\alpha = 1$.

---

Definition:

$$
y_t = y_{t-1} + e_t
$$

---

Interpretation

- Variance increases **linearly with time**
- No mean reversion → strong trending behavior
- Nonstationary time series
- Best prediction of next value is the **current value**
- Random walk is a **martingale**
